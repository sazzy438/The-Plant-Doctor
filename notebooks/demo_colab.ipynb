# Plant Doctor — free public demo link (Google Colab)

This notebook runs Plant Doctor on a free Google Colab machine and gives you a **public link** you can send to anyone. The link works while this notebook stays open (Colab stops idle notebooks after a while, and free sessions last at most about 12 hours), so it is for demos, not permanent hosting.

**How to use:** choose **Runtime → Run all**, wait 3–5 minutes, then click the link that ends in `.gradio.live` printed at the bottom. No GPU is needed; leave the runtime type on CPU.

```python
#@title 1. Get the project (from GitHub, or upload the ZIP if the repository is private)
GITHUB_REPO = "https://github.com/sazzy438/The-Plant-Doctor.git"  #@param {type:"string"}
import os, subprocess, shutil, zipfile, pathlib
if os.path.exists('/content/plant-doctor'):
    shutil.rmtree('/content/plant-doctor')
r = subprocess.run(["git", "clone", "--depth", "1", GITHUB_REPO, "/content/plant-doctor"], capture_output=True, text=True)
if r.returncode != 0:
    print("Could not clone the repository (it may be private). Please upload plant-doctor.zip now:")
    from google.colab import files
    up = files.upload()
    zname = next(iter(up))
    with zipfile.ZipFile(zname) as z:
        z.extractall('/content')
    inner = next(p for p in pathlib.Path('/content').iterdir() if (p / 'app.py').exists())
    if str(inner) != '/content/plant-doctor':
        inner.rename('/content/plant-doctor')
assert os.path.exists('/content/plant-doctor/app.py'), "app.py not found - check the repository URL or ZIP"
assert os.path.exists('/content/plant-doctor/artifacts/plant_doctor_efficientnet_b0.pt'), "trained model file missing"
print("Project ready")
```

```python
#@title 2. Install the few packages Colab does not already have (about 1 minute)
import subprocess, sys
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "gradio>=6,<7", "reportlab>=4", "scipy", "scikit-learn", "matplotlib", "pillow"], check=True)
import torch, gradio; print("torch", torch.__version__, "| gradio", gradio.__version__)
```
